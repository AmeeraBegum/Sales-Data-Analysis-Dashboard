## Step 6: Create Rounded Dashboard Panels

```python
def rounded_panel(
    x,
    y,
    width,
    height,
    edge="#C9CED5",
    face="#FCFCFC",
    linewidth=1.0,
    radius=0.015
):

    panel = FancyBboxPatch(

        (x, y),

        width,

        height,

        boxstyle=f"round,pad=0.003,rounding_size={radius}",

        transform=fig.transFigure,

        facecolor=face,

        edgecolor=edge,

        linewidth=linewidth,

        zorder=-10
    )

    fig.patches.append(panel)
```
